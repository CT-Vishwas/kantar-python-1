### Handling missing values

```python
d2 = {
    "col1":[None,1,2,3],
    "col2":[1,None,None,3],
    "col3":[1,2,None,20]
}
df1 = pd.DataFrame(d2)
print(df1.isna().sum())
print(df1)
df2 = df1.dropna()
print(df1)
df2.isna().sum()
print(df2)
```

### Adding new Rows of data

```python
new_row = ["Cady", 27, "Chandigarh", 23000]
df.loc[len(df)] = new_row
print(df)
```

```python
new_rows = [
    {"names": "Balram", "age": 45,"city":"Kolkata", "salary":90000},
    {"names": "Jane", "age": 23,"city":"Kolkata", "salary":90000},
    {"names": "Jugnu", "age": 46,"city":"Kolkata", "salary":8000}
]

new_df = pd.DataFrame(new_rows)

df = pd.concat([df, new_df], ignore_index=True)
print(df)
```

### Converting dtypes

```python
df["salary"] = df["salary"].astype(float)
```

```python
print(df)
```

### Removing Duplicates

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df)
```

### Renaming the columns

```python
df.columns = df.columns.str.upper()
print(df)
```

### Filtering data to new dataframe

```python
new_df = df[df["AGE"] > 21]
```

```python
print(new_df)
```

```python
new_df = df.loc[df["CITY"] == "Unkown", ["NAMES","AGE"]]
print(new_df)
```

### Creating New Columns

```python
df["age_group"] = df["AGE"].apply(
    lambda x: "Young" if x < 25 else "Adult"
)
print(df)
```

```python
df["salary_bonus"] = df["SALARY"] * 0.1
print(df)
```

### Grouping of Data

```python
# average salary based on city
df.groupby('CITY')['SALARY'].mean()
```

```python
df.groupby('CITY').agg({
    'salary_bonus': 'sum',
    'SALARY': 'mean',
    'AGE':'mean'
})
```

### Merging or Joining of data

```python
d3 = {
    "NAMES": ["vishwas","arjun","ramya","John","Krishna","Cady","Balram","Jane","Jugnu"],
    "DEPARTMENT": ["IT","SALES","IT","SALES","FINANCE","FINANCE","SALES","SALES","IT"]
}

new_df = pd.DataFrame(d3)
df_joined = pd.merge(df,new_df, on="NAMES", how="inner") # left,right,outer
print(df_joined)
```

```python
df_pivot = df_joined.pivot_table(
    values = "salary_bonus",
    index = "DEPARTMENT",
    columns="CITY",
    aggfunc = "sum"
)

df_pivot.fillna(0, inplace=True)
#print(df_pivot)
df_pivot.head()
```

```python
assert df_joined['AGE'].max() < 60
```

### Connecting with a context manager

```python
from sqlalchemy import create_engine, Integer, String, Numeric
engine = create_engine("sqlite:///employee_data.db")

with engine.begin() as conn:
    df_joined.to_sql(
        "employees",
        con=conn,
        if_exists = "replace",
        index=False,
        dtype={
            "NAMES":String(100),
            "AGE": Integer,
            "CITY": String(100),
            "SALARY": Numeric(8,2),
            "age_group":String(100),
            "salary_bonus":Numeric(8,2),
            "DEPARTMENT":String(100)
        }
    )
```

```python
df_db = pd.read_sql("SELECT * FROM employees", engine)
print(df_db)
```
