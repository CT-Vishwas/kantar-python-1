## First Exploring Pandas

```python
d1 = {
"names": ["vishwas","arjun","ramya","John","Krishna"],
"city": ["Bangalore","mumbai","pune",None,"Kolkata"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(d1)
```

### Adding Columns to a Dataframe

```python
df['salary'] = [25000,45000,75000,90000,10000]
```

```python
print(df)
```

```python
ages = [25,21,19,35,10]
df.insert(loc=1, column="age", value=ages)
print(df)
```

### Handling missing values

```python
d2 = {
    "col1":[None,1,2,3],
    "col2":[1,None,None,3],
    "col3":[1,2,None,20]
}
df1 = pd.DataFrame(d2)
print(df1.isna().sum())
print(df1)
df2 = df1.dropna()
print(df1)
df2.isna().sum()
print(df2)
```

```python
df["city"].fillna("Unkown", inplace=True)
```

```python
print(df)
```

### Adding new Rows of data

```python
new_row = ["Cady", 27, "Chandigarh", 23000]
df.loc[len(df)] = new_row
print(df)
```

```python
new_rows = [
    {"names": "Balram", "age": 45,"city":"Kolkata", "salary":90000},
    {"names": "Jane", "age": 23,"city":"Kolkata", "salary":90000},
    {"names": "Jugnu", "age": 46,"city":"Kolkata", "salary":8000}
]

new_df = pd.DataFrame(new_rows)

df = pd.concat([df, new_df], ignore_index=True)
print(df)
```

### Converting dtypes

```python
df["salary"] = df["salary"].astype(float)
```

```python
print(df)
```

### Removing Duplicates

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df)
```

### Renaming the columns

```python
df.columns = df.columns.str.upper()
print(df)
```

### Filtering data to new dataframe

```python
new_df = df[df["AGE"] > 21]
```

```python
print(new_df)
```

```python
new_df = df.loc[df["CITY"] == "Unkown", ["NAMES","AGE"]]
print(new_df)
```
