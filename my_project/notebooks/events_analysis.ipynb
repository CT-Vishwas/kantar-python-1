Merging Events & Pricing

Calculating the Effective price 

```python
event_pricing["effective_price"] = event_pricing["base_price"] * (1 - (event_pricing["discount_pct"]/100))
```

Converting datetime types

```python
event_pricing["scheduled_date_time"] = pd.to_datetime(event_pricing["scheduled_date_time"])
```

```python
registrations_attendees["reg_timestamp"] = pd.to_datetime(registrations_attendees["reg_timestamp"])
```

How many unique values are there in few categorical columns

```python
registrations_attendees['role'].nunique()
```

```python
registrations_attendees['industry'].nunique()
```

We will represent the categorical values as a category dtype for memory efficiency

```python
columns_to_convert = ['payment_status', 'ticket_type','city','role','industry']
for col in columns_to_convert:
    registrations_attendees[col] = registrations_attendees[col].astype('category')
```
